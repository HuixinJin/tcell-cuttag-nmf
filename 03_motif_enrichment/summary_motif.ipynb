## H3K27ac

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import numpy as np
```

```python
dataframes = []
for h in ['H3K27ac', 'H3K4me1', 'H3K4me3']:
    motif_all = pd.DataFrame()
    if h == 'H3K4me3':
        n = 10
    else:
        n = 16
    for i in range(1, n):
        file_path = glob.glob('./' + h + '/meme-chip-online/cluster' + str(i) + '/*/summary.tsv')
        motif = pd.read_table(file_path[0])
        motif = motif.iloc[0:motif.shape[0] - 3]
        if motif.shape[0] == 0:
            continue
        '''
        motif['motif'] = np.where(
            motif['MOTIF_SOURCE'].str.contains('JASPAR', na=False),
            motif['MOTIF_ID'] + '_' + motif['ALT_ID'],
            motif['MOST_SIMILAR_MOTIF']
        )
        '''
        motif['cluster'] = i
        motif = motif[['cluster'] + [col for col in motif.columns if col != 'cluster']]
        motif_all = pd.concat([motif_all, motif], axis=0)
    dataframes.append(motif_all)
```

```python
sheet_names = ['H3K27ac', 'H3K4me1', 'H3K4me3']
output_file = 'cut_tag_cluster_motif_enrichment.xlsx'

with pd.ExcelWriter(output_file) as writer:
    for df, sheet_names in zip(dataframes, sheet_names):
        df.to_excel(writer, sheet_name=sheet_names, index=False)

print("DataFrames written to Excel file successfully.")
```

```python
dataframes = []
#for h in ['H3K27ac', 'H3K4me1', 'H3K4me3', 'H3K9me3']:
for h in ['H3K9me3', 'H3K9me3_rank3']:
    motif_all = pd.DataFrame()
    if h == 'H3K4me3' or h == 'H3K9me3' :
        n = 10
    elif h == 'H3K9me3_rank3':
        n = 4
    else:
        n = 16
    for i in range(1, n):
        file_path = glob.glob('./homer/' + h + '/cluster_' + str(i) + '/knownResults.txt')
        motif = pd.read_table(file_path[0])
        if motif.shape[0] == 0:
            continue
        '''
        motif['motif'] = np.where(
            motif['MOTIF_SOURCE'].str.contains('JASPAR', na=False),
            motif['MOTIF_ID'] + '_' + motif['ALT_ID'],
            motif['MOST_SIMILAR_MOTIF']
        )
        '''
        motif = motif[motif['q-value (Benjamini)'] < 0.05]
        motif.columns = motif.columns[0:5].tolist() + ['# of Target Sequences with Motif',
        r'% of Target Sequences with Motif',
        r'# of Background Sequences with Motif',
        r'% of Background Sequences with Motif']
        motif['cluster'] = i
        motif = motif[['cluster'] + [col for col in motif.columns if col != 'cluster']]
        motif_all = pd.concat([motif_all, motif], axis=0)
    dataframes.append(motif_all)
```

```python
sheet_names = ['H3K9me3', 'H3K9me3_rank3']
output_file = 'homer_H3K9me3_cut_tag_cluster_motif_enrichment.xlsx'

with pd.ExcelWriter(output_file) as writer:
    for df, sheet_names in zip(dataframes, sheet_names):
        df.to_excel(writer, sheet_name=sheet_names, index=False)

print("DataFrames written to Excel file successfully.")
```
