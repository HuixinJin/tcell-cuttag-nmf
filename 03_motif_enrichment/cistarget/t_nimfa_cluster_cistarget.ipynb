```python
# Load cistarget functions
from pycistarget.motif_enrichment_cistarget import *
import pyranges as pr
import os
import glob
import pandas as pd
```

```python
enrich_out = []
for h in ["H3K27ac","H3K4me1","H3K4me3"]:
    #path_to_region_sets = '../t/result/H3K27ac/cluster_bed/'
    region_sets_files = glob.glob('../t/result/' + str(h) + '/cluster_bed/cluster_*_peaks.bed')
    region_sets = {os.path.basename(x).replace('.bed', ''):pr.read_bed(x) for x in region_sets_files}
    # Run, using precomputed database
    cistarget_dict = run_cistarget(ctx_db = '/dshare/home/jinhx/database/pycistarget/mm10/mm10_screen_v10_clust.regions_vs_motifs.rankings.feather',
                                                          region_sets = region_sets,
                                                          specie = 'mus_musculus',
                                                          auc_threshold = 0.005,
                                                          #nes_threshold = 3.0,
                                                          nes_threshold = 2.0,
                                                          rank_threshold = 0.05,
                                                          #annotation = ['Direct_annot', 'Orthology_annot'],
                                                          annotation_version = 'v10nr_clust',
                                                          path_to_motif_annotations = '/dshare/home/jinhx/database/pyscenic/mus/motifs-v10nr_clust-nr.mgi-m0.001-o0.0.tbl',
                                                          n_cpu = 1)
    out_df = pd.DataFrame()
    for x in region_sets:
        output_html = str(h) + '/' + x + '_motif_enricment.html'
        out_df = pd.concat([out_df, cistarget_dict[x].motif_enrichment])
        cistarget_dict[x].motif_enrichment.to_html(open(output_html, 'w'), escape=False, col_space=80)
    enrich_out.append(out_df)
```

```python
len(enrich_out)
```

```python
sheet_names = ['H3K27ac', 'H3K4me1', 'H3K4me3']
output_file = 'cistarget_cut_tag_cluster_motif_enrichment.xlsx'

with pd.ExcelWriter(output_file) as writer:
    for df, sheet_names in zip(enrich_out, sheet_names):
        df.to_excel(writer, sheet_name=sheet_names, index=False)
```
