```python
with open('../anno_chipseeker/H3K9me3/chipseeker_cluster_2_anno_unique_genes_name.txt', 'r') as f:
    chipseeker_genes = [line.strip() for line in f]

with open('H3K9me3/cluster_2_anno_unique_genes_name.txt', 'r') as f:
    homer_genes = [line.strip() for line in f]

```

```python
# Create sets from lists
chipseeker_genes = set(chipseeker_genes)
homer_genes = set(homer_genes)

# Create Venn diagram
venn2([chipseeker_genes, homer_genes], ('chipseeker', 'homer'))

# Add labels
plt.title('Venn Diagram of Two Lists')
plt.show()
```

```python
# Load all gene sets
all_genes = {}
for i in range(1, 10):
    df = pd.read_table(f'../anno_chipseeker/H3K9me3/peak_annotation_{i}.tsv')
    genes = set(df['SYMBOL'].dropna())
    all_genes[f'cluster_{i}'] = genes

# Get genes unique to cluster 2
cluster_2_genes = all_genes['cluster_2']
other_clusters_genes = set()

# Combine genes from all other clusters
for i in range(1, 10):
    if i != 2:
        other_clusters_genes.update(all_genes[f'cluster_{i}'])

# Find genes unique to cluster 2
unique_to_cluster_2 = cluster_2_genes - other_clusters_genes

# Convert to list if needed
unique_to_cluster_2_list = list(unique_to_cluster_2)

# Print the result
with open('../anno_chipseeker/H3K9me3/chipseeker_only_in_cluster_2_anno_gene_name.txt', 'w') as f:
    f.write('\n'.join(unique_to_cluster_2))
```

```python
# Create a dictionary to store all gene sets
gene_sets = {}

# Load all 9 datasets
for i in range(1, 10):
    df = pd.read_table('../anno_chipseeker/H3K9me3/peak_annotation_' + str(i) + '.tsv')
    genes = df['SYMBOL'].dropna()  # Drop any NaN values
    gene_sets[f'Cluster_{i}'] = set(genes)

# Convert to a format suitable for upsetplot
from upsetplot import from_contents
upset_data = from_contents(gene_sets)

# Create the upset plot
plt.figure(figsize=(12, 8))
upsetplot.plot(upset_data, show_counts=True)
plt.suptitle('UpSet Plot of Gene Overlaps Across 9 Clusters')
plt.show()
```
