```python
import pandas as pd
import os
from openpyxl import Workbook
```

```python

def txt_to_excel(directory, output_file):
    # Create a new Excel workbook
    wb = Workbook()
    
    # Remove default sheet
    wb.remove(wb.active)
    
    # Get all txt files in the directory
    txt_files = [f for f in os.listdir(directory) if f.endswith('.txt')]
    
    for txt_file in txt_files:
        # Get sheet name by splitting at "_eGO.txt"
        sheet_name = txt_file.split('_eGO.txt')[0]
        
        # Read the txt file
        df = pd.read_csv(os.path.join(directory, txt_file), sep='\t')
        
        # Create a new sheet with the txt file name
        ws = wb.create_sheet(title=sheet_name)
        
        # Write column headers
        for col_idx, column in enumerate(df.columns, 1):
            ws.cell(row=1, column=col_idx, value=column)
        
        # Write data
        for row_idx, (index, row) in enumerate(df.iterrows(), 2):
            for col_idx, value in enumerate(row, 1):
                ws.cell(row=row_idx, column=col_idx, value=value)
    
    # Save the workbook
    wb.save(output_file)
    print(f"Excel file saved as {output_file}")

```

```python
# Example usage
directory = 'GO_enrich/H3K4me1'  # Replace with your directory
output_file = 'eGO_list.xlsx'  # Replace with your desired output file name
txt_to_excel(directory, output_file)
```
